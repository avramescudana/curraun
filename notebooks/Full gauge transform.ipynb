# Perform the gauge transformation without approximation

Set the relevant parameters and variables

```python
import numpy as np

# hbar * c [GeV * fm]
hbarc = 0.197326 

# Simulation box 
L = 2.5         
N = 128 
tau_sim = 1.25    
DTS = 8

# Derived parameters
a = L/N
E0 = N / L * hbarc
DT = 1.0 / DTS
maxt = int(tau_sim / a * DTS)
nplus = maxt//DTS     

# Glasma fields
su_group = 'su3'
Qs = 2        
ns = 50    
factor = 0.68        
g2mu = Qs / factor     
g = np.pi * np.sqrt(1 / np.log(Qs / 0.2))          		
mu = g2mu / g**2          	
ir = 0.2 * g**2 * mu         
uv = 10.0       


nevents = 1
```

```python
import os

os.environ["MY_NUMBA_TARGET"] = "numba"
os.environ["PRECISION"] = "double"
os.environ['GAUGE_GROUP'] = su_group

# Import relevant modules
import sys
sys.path.append('..')

# Glasma modules
import curraun.core as core
import curraun.mv as mv
import curraun.initial as initial
initial.DEBUG = False

import curraun.su as su
from curraun.numba_target import use_cuda
if use_cuda:
    from numba import cuda

import curraun.su as su
```

Define the simulation routine

```python
from tqdm import tqdm 

# Simulation routine
def simulate():
    
    # Dictionary to store the results
    output = {}
    
    # Derived parameters
    a = L/N
    E0 = N/L*hbarc
    DT = 1.0/DTS 
    maxt = int(tau_sim/a*DTS)
    
    # We create the object Glasma Simulation
    s = core.Simulation(N, DT, g)
    
    # We intialize the Glasma fields
    va = mv.wilson(s, mu=mu/E0, m=ir/E0, uv=uv/E0, num_sheets=ns)
    vb = mv.wilson(s, mu=mu/E0, m=ir/E0, uv=uv/E0, num_sheets=ns)
    initial.init(s, va, vb)
    
    # We create an object to store the fields at every time step
    ux_sim = np.zeros((maxt, N+1, N+1, su.GROUP_ELEMENTS), dtype=su.GROUP_TYPE)
    aeta_sim = np.zeros((maxt, N+1, N+1, su.GROUP_ELEMENTS), dtype=su.GROUP_TYPE)
    
    # We create objects to store the plus links links
    up_temp = np.zeros((maxt//DTS, N+1, N+1, su.GROUP_ELEMENTS), dtype=su.GROUP_TYPE)
    
    # We create an object to store the gauge transformation operator
    vlc_dag = np.zeros((maxt//DTS, N+1, N+1, su.GROUP_ELEMENTS), dtype=su.GROUP_TYPE)
    
    # We create an object to store the plus links in lc gauge
    up_lc = np.zeros((maxt//DTS-1, N+1, N+1, su.GROUP_ELEMENTS), dtype=su.GROUP_TYPE)
    
    # We create an object to reorder the z coordinate in a suitable way for Meijian's code
    up_reorder = np.zeros((maxt//DTS-1, N+1, N+1, su.GROUP_ELEMENTS), dtype=su.GROUP_TYPE)
    
    # We set them to unit
    for xplus in range(maxt//DTS):
        for y in range(N):
            for z in range(N):
                up_temp[xplus, y, z] = su.unit()
    
    # We first simulate the Glasma event and store the fields at every time step
    for t in range(maxt):
        
        # Evolve the Glasma fields
        core.evolve_leapfrog(s)
        
        # Store the fields at every time step
        u1 = s.u1.copy()
        ux_sim[t] = u1[:, 0, :].reshape(N, N, su.GROUP_ELEMENTS)
        
        aeta = s.aeta1.copy()
        aeta_sim[t] = aeta.reshape(N, N, su.GROUP_ELEMENTS)
    
    
    # We now perform the gauge transformation to the lc gauge
    with tqdm(total=maxt//DTS) as pbar:
        for xplus in range(maxt//DTS):
                
            for y in range(N):
                for z in range(N):
                    
                    # We construct the plus link in LC coordinates from the links in Milne coordinates
                    if -xplus<(z-N//2)<xplus:
                        tau = round(DTS*np.sqrt(xplus**2 - (z-N//2)**2))
                        
                        aeta_fact = su.mul_s(aeta_sim[tau, xplus+1, y], -(z-N/2)/tau**2)
                        ut = su.mexp(aeta_fact)
                        
                        up_temp[xplus, y, z] = su.mul(ux_sim[tau, xplus, y], ut)
                        
                    # We construct the gauge transformation operator
                    vlc0 = su.unit(); vlc1 = su.unit()
                    for xminus in range(0, -xplus, -1):
                        if -(xplus+xminus-1)<(z-N//2)<(xplus+xminus-1):
                            tau = round(DTS*np.sqrt((xplus+xminus-1)**2 - (z-N//2)**2))
                            
                            aeta_fact = su.mul_s(aeta_sim[tau, xplus-xminus, y], -(z-N/2)/tau**2)
                            ut = su.mexp(aeta_fact)
                            
                            umin_dag = su.mul(su.dagger(ut), ux_sim[tau, xplus-xminus, y])
                            
                            vlc1 = su.mul(vlc0, umin_dag)
                            vlc0 = np.copy(vlc1)
                    vlc_dag[xplus, y, z] = vlc1

                    # We perform the gauge transformation of the field at the previous time step
                    if xplus != 0:
                        aux = su.mul(up_temp[(xplus-1), y, z], vlc_dag[xplus, y, z])
                        aux2 = su.mul(su.dagger(vlc_dag[(xplus-1), y, z]), aux)
                        up_lc[(xplus-1), y, z] = su.dagger(aux2)
                    
                    # We reorder the fields in a suitable format to input in Meijian's code
                    if z-N/2 >= 0:
                        up_reorder[xplus-1, y, z-N//2] = up_lc[xplus-1, y, z]
                            
                    else:
                        up_reorder[xplus-1, y, N//2+z] = up_lc[xplus-1, y, z]
                            
            pbar.update(1)
            
    if use_cuda:
        cuda.current_context().deallocations.clear()
    
    up_jetsim = up_reorder.reshape(maxt//DTS-1, N*N, su.GROUP_ELEMENTS)
    
    output['up_temp'] = up_temp
    output['up_lc'] = up_lc
    output['up_jetsim'] = up_jetsim
    
    return output
```

```python
# We perform the simulation
import warnings
warnings.filterwarnings('ignore')

for n in range (nevents):
    
    output = simulate()
    
    up_temp = output['up_temp']
    up_lc = output['up_lc']
    up_jetsim = output['up_jetsim']
    
    # np.save(os.path.join('simulations', 'ux_lc', 'result_%i.npy' %n), up_jetsim)
```

Let us now check the gauge transformation with the plaquettes

```python
from tqdm import tqdm 

# Simulation routine
def simulate():
    
    # Dictionary to store the results
    output = {}
    
    # Derived parameters
    a = L/N
    E0 = N/L*hbarc
    DT = 1.0/DTS 
    maxt = int(tau_sim/a*DTS)
    
    # We create the object Glasma Simulation
    s = core.Simulation(N, DT, g)
    
    # We intialize the Glasma fields
    va = mv.wilson(s, mu=mu/E0, m=ir/E0, uv=uv/E0, num_sheets=ns)
    vb = mv.wilson(s, mu=mu/E0, m=ir/E0, uv=uv/E0, num_sheets=ns)
    initial.init(s, va, vb)
    
    # We create an object to store the fields at every time step
    ux_sim = np.zeros((maxt, N, N, su.GROUP_ELEMENTS), dtype=su.GROUP_TYPE)
    aeta_sim = np.zeros((maxt, N, N, su.GROUP_ELEMENTS), dtype=su.GROUP_TYPE)
    
    # We create objects to store the gauge links
    up_temp = np.zeros((2, maxt//DTS-1, N, N, su.GROUP_ELEMENTS), dtype=su.GROUP_TYPE)
    um_temp = np.zeros((maxt//DTS-1, N, N, su.GROUP_ELEMENTS), dtype=su.GROUP_TYPE)
    
    up_lc = np.zeros((2, maxt//DTS-1, N, N, su.GROUP_ELEMENTS), dtype=su.GROUP_TYPE)
    
    # We create an object to store the gauge transformation operator
    vlc_dag = np.zeros((2, maxt//DTS, N, N, su.GROUP_ELEMENTS), dtype=su.GROUP_TYPE)
    
    # We creaate an object to store the plaquette
    upm_temp = np.zeros((maxt//DTS-2, N, N, su.GROUP_ELEMENTS), dtype=su.GROUP_TYPE)
    upm_lc = np.zeros((maxt//DTS-2, N, N, su.GROUP_ELEMENTS), dtype=su.GROUP_TYPE)
    
    # We create an object to store the gauge invariant
    res_temp = np.zeros((maxt//DTS-2, N, N), dtype=su.GROUP_TYPE)
    res_lc = np.zeros((maxt//DTS-2, N, N), dtype=su.GROUP_TYPE)
    res = np.zeros((maxt//DTS-2, N, N), dtype=su.GROUP_TYPE)
    
    # We set them to unit
    for xplus in range(maxt//DTS-1):
        for y in range(N):
            for z in range(N):
                up_temp[0, xplus, y, z] = su.unit()
                up_temp[1, xplus, y, z] = su.unit()
                um_temp[xplus, y, z] = su.unit()
    
    # We first simulate the Glasma event and store the fields at every time step
    for t in range(maxt):
        
        # Evolve the Glasma fields
        core.evolve_leapfrog(s)
        
        # Store the fields at every time step
        u1 = s.u1.copy()
        ux_sim[t] = u1[:, 0, :].reshape(N, N, su.GROUP_ELEMENTS)
        
        aeta1 = s.aeta1.copy()
        aeta_sim[t] = aeta1.reshape(N, N, su.GROUP_ELEMENTS)
    
    
    # We now perform the gauge transformation to the lc gauge at x^- = 0 and x^- = 1
    with tqdm(total=maxt//DTS-1) as pbar:
        for xplus in range(maxt//DTS-1):
                
            for y in range(N):
                for z in range(N):
                    
                    # We construct the plus and minus links at x^- = 0
                    if -xplus<(z-N//2)<xplus:
                        tau = round(DTS*np.sqrt(xplus**2 - (z-N//2)**2))
                        
                        aeta_fact1 = su.mul_s(aeta_sim[tau, xplus+1, y], -(z-N/2)/tau**2)
                        ut1 = su.mexp(aeta_fact1)
                        
                        up_temp[0, xplus, y, z] = su.mul(ux_sim[tau, xplus, y], ut1)
                        
                        aeta_fact2 = su.mul_s(aeta_sim[tau, xplus-1, y], -(z-N/2)/tau**2)
                        ut2 = su.mexp(aeta_fact2)
                        
                        um_temp[xplus, y, z] = su.mul(su.dagger(ux_sim[tau, xplus-1, y]), ut2)
                        
                    
                    # We construct the plus links at x^- = 1
                    if -(xplus+1)<(z-N//2)<(xplus+1):
                        tau = round(DTS*np.sqrt((xplus+1)**2 - (z-N//2)**2))
                        
                        aeta_fact = su.mul_s(aeta_sim[tau, xplus, y], -(z-N/2)/tau**2)
                        ut = su.mexp(aeta_fact)
                        
                        up_temp[1, xplus, y, z] = su.mul(ux_sim[tau, xplus-1, y], ut)
                        
                    
                    # We construct the gauge transformation operator at x^- = 0
                    vlc0 = su.unit(); vlc1 = su.unit()
                    for xminus in range(0, -xplus, -1):
                        if -(xplus+xminus-1)<(z-N//2)<(xplus+xminus-1):
                            tau = round(DTS*np.sqrt((xplus+xminus-1)**2 - (z-N//2)**2))
                            
                            aeta_fact = su.mul_s(aeta_sim[tau, xplus-xminus, y], -(z-N/2)/tau**2)
                            ut = su.mexp(aeta_fact)
                            
                            umin_dag = su.mul(su.dagger(ut), ux_sim[tau, xplus-xminus, y])
                            
                            vlc1 = su.mul(vlc0, umin_dag)
                            vlc0 = np.copy(vlc1)
                    
                    vlc_dag[0, xplus, y, z] = vlc1
                    
                    # We construct the gauge transformation operator at x^- = 1
                    vlc0 = su.unit(); vlc1 = su.unit()
                    for xminus in range(1, -xplus, -1):
                        if -(xplus+xminus-1)<(z-N//2)<(xplus+xminus-1):
                            tau = round(DTS*np.sqrt((xplus+xminus-1)**2 - (z-N//2)**2))
                            
                            aeta_fact = su.mul_s(aeta_sim[tau, xplus-xminus, y], -(z-N/2)/tau**2)
                            ut = su.mexp(aeta_fact)
                            
                            umin_dag = su.mul(su.dagger(ut), ux_sim[tau, xplus-xminus, y])
                            
                            vlc1 = su.mul(vlc0, umin_dag)
                            vlc0 = np.copy(vlc1)
                    
                    vlc_dag[1, xplus, y, z] = vlc1
                    
                    # We perform the gauge transformation of the field at the previous time step at x^- = 0
                    if xplus != 0:
                        aux = su.mul(up_temp[0, (xplus-1), y, z], vlc_dag[0, xplus, y, z])
                        up_lc[0, xplus-1, y, z] = su.mul(su.dagger(vlc_dag[0, (xplus-1), y, z]), aux)
                    
                    # We perform the gauge transformation of the field at the previous time step at x^- = 1
                    if xplus != 0:
                        aux = su.mul(up_temp[1, (xplus-1), y, z], vlc_dag[1, xplus, y, z])
                        up_lc[1, xplus-1, y, z] = su.mul(su.dagger(vlc_dag[1, (xplus-1), y, z]), aux)
                    
                            
            pbar.update(1)
    
    
    with tqdm(total=maxt//DTS-2) as pbar:
        for xplus in range(maxt//DTS-2):
            for y in range(N):
                for z in range(N):
                
                    # We compute the plaquette in temp gauge
                    aux1 = su.mul(up_temp[0, xplus, y, z], um_temp[xplus+1, y, z])
                    aux2 = su.mul(su.dagger(up_temp[1, xplus, y, z]), su.dagger(um_temp[xplus, y, z]))
                
                    upm_temp[xplus, y, z] = su.mul(aux1, aux2)
                    
                    # We compute the gauge invariant
                    s1 = su.mul_s(su.unit(), 2)
                    s2 = su.mul_s(upm_temp[xplus, y, z], -1)
                    s3 = su.mul_s(su.dagger(upm_temp[xplus, y, z]), -1)
                    
                    s4 = su.add(s1, s2)
                    s5 = su.add(s4, s3)
                    
                    res_temp[xplus, y, z] = su.tr(s5)
                    
                    
                    # We compute the plaquette in lc gauge
                    upm_lc[xplus, y, z] = su.mul(up_lc[0, xplus, y, z], su.dagger(up_lc[1, xplus, y, z]))
                    
                    # We compute the gauge invariant
                    s1 = su.mul_s(su.unit(), 2)
                    s2 = su.mul_s(upm_lc[xplus, y, z], -1)
                    s3 = su.mul_s(su.dagger(upm_lc[xplus, y, z]), -1)
                    
                    s4 = su.add(s1, s2)
                    s5 = su.add(s4, s3)
                    
                    res_lc[xplus, y, z] = su.tr(s5)
                    
                    # We compute the subtraction of both gauge invariants
                    res[xplus, y, z] = res_temp[xplus, y, z] - res_lc[xplus, y, z]

            pbar.update(1)
    
    
    if use_cuda:
        cuda.current_context().deallocations.clear()
    
    
    output['res_temp'] = res_temp
    output['res_lc'] = res_lc
    output['res'] = res
    

    return output
```

```python
# We perform the simulation
import warnings
warnings.filterwarnings('ignore')

output = simulate()

res_temp = output['res_temp']
res_lc = output['res_lc']
res = output['res']
```
