# Nice plots for the paper

## 1. Momentum broadening

### 1.1 Method comparison

Load the quantum results, both for the Direct and Lorentz force calculation

```python
# We load the direct result
import numpy as np

xp_direct = np.loadtxt('paper_plots/direct_pkin.txt')[0]
delta_pkin_y2_direct = np.loadtxt('paper_plots/direct_pkin.txt')[1]
delta_pkin_z2_direct = np.loadtxt('paper_plots/direct_pkin.txt')[2]
```

```python
# We load the Lorentz result

xp_Lorentz = np.loadtxt('paper_plots/Lorentz_pkin.txt')[0]
delta_pkin_y2_Lorentz = np.loadtxt('paper_plots/Lorentz_pkin.txt')[1]
delta_pkin_z2_Lorentz = np.loadtxt('paper_plots/Lorentz_pkin.txt')[2]
```

```python
# We load the classical result

xp_class = np.loadtxt('classical_results_g2muL=100/classical_pi.txt')[0]
delta_pkin_y2_class = np.loadtxt('classical_results_g2muL=100/classical_pi.txt')[1]
delta_pkin_z2_class = np.loadtxt('classical_results_g2muL=100/classical_pi.txt')[2]
```

Plot the method comparison results

```python
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# ----------------------------
# Paper-style rcParams
# ----------------------------
mpl.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman", "Times", "STIXGeneral", "DejaVu Serif"],
    "mathtext.fontset": "stix",
    "font.size": 9,

    "axes.labelsize": 9,
    "axes.titlesize": 9,
    "axes.linewidth": 0.9,

    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.major.size": 5,
    "ytick.major.size": 5,
    "xtick.minor.size": 3,
    "ytick.minor.size": 3,
    "xtick.major.width": 0.9,
    "ytick.major.width": 0.9,
    "xtick.minor.width": 0.8,
    "ytick.minor.width": 0.8,

    "xtick.labelsize": 8,
    "ytick.labelsize": 8,
    "legend.fontsize": 8,
    "legend.frameon": False,
    "savefig.dpi": 300,
})

# ----------------------------
# Constants
# ----------------------------
shift = 50 / 1024 * 2
conv  = 2 / 0.197326

c_y, c_z = "tab:red", "tab:blue"

ms = 4.2
mew = 0.8
lw = 1.3

# ----------------------------
# Figure
# ----------------------------
fig, ax = plt.subplots(figsize=(3.39, 2.6))   # nice single-column journal size

# Direct
ax.plot(xp_direct+shift, delta_pkin_y2_direct, linestyle="None",
        marker="s", markersize=ms, markeredgewidth=mew, markerfacecolor="none",
        color=c_y)
ax.plot(xp_direct+shift, delta_pkin_z2_direct, linestyle="None",
        marker="s", markersize=ms, markeredgewidth=mew, markerfacecolor="none",
        color=c_z)

# Lorentz
ax.plot(xp_Lorentz, delta_pkin_y2_Lorentz, linestyle="None",
        marker="o", markersize=ms, markeredgewidth=mew, markerfacecolor="none",
        color=c_y)
ax.plot(xp_Lorentz, delta_pkin_z2_Lorentz, linestyle="None",
        marker="o", markersize=ms, markeredgewidth=mew, markerfacecolor="none",
        color=c_z)

# Classical
ax.plot(xp_class * conv, delta_pkin_y2_class / 3, linestyle="--", linewidth=lw, color=c_y)
ax.plot(xp_class * conv, delta_pkin_z2_class / 3, linestyle="--", linewidth=lw, color=c_z)

# Labels
ax.set_xlabel(r"$x^+\ (\mathrm{GeV}^{-1})$")
ax.set_ylabel(r"$\langle (\delta p^{\mathrm{kin}}_{i})^2 \rangle\ (\mathrm{GeV}^2)$")

# Ticks & grid
ax.minorticks_on()
ax.grid(True, which="major", linewidth=0.5, alpha=0.25)
ax.grid(True, which="minor", linewidth=0.3, alpha=0.15)
ax.tick_params(top=True, right=True)

ax.set_xlim(left=0)
ax.margins(x=0.02, y=0.05)

# ----------------------------
# Two legends
# ----------------------------
legend_method = [
    Line2D([0], [0], marker="s", linestyle="None", markersize=ms,
           markerfacecolor="none", markeredgewidth=mew,
           color="black", label="Direct"),

    Line2D([0], [0], marker="o", linestyle="None", markersize=ms,
           markerfacecolor="none", markeredgewidth=mew,
           color="black", label="Lorentz"),

    Line2D([0], [0], linestyle="--", linewidth=lw,
           color="black", label="Classical"),
]

legend_comp = [
    Line2D([0], [0], color=c_y, lw=lw, label=r"$i=y$"),
    Line2D([0], [0], color=c_z, lw=lw, label=r"$i=z$"),
]

leg1 = ax.legend(handles=legend_method, loc="upper left", title="Method")
ax.add_artist(leg1)

ax.legend(handles=legend_comp, loc="lower right", title="Component")

# ----------------------------
# Save
# ----------------------------
fig.tight_layout()
fig.savefig("paper_plots/Momentum_broadening_comparison.pdf", bbox_inches="tight")
plt.show()
```

### 1.2 Width dependence

For the $y$ component we load the results

```python
# We load the direct result
import numpy as np

delta_pkin_y2_wy5 = np.loadtxt('paper_plots/pkin_wy5.txt')[1]
delta_pkin_z2_wy5 = np.loadtxt('paper_plots/pkin_wy5.txt')[2]

delta_pkin_y2_wy10 = np.loadtxt('paper_plots/pkin_wy10.txt')[1]
delta_pkin_z2_wy10 = np.loadtxt('paper_plots/pkin_wy10.txt')[2]

delta_pkin_y2_wy20 = np.loadtxt('paper_plots/pkin_wy20.txt')[1]
delta_pkin_z2_wy20 = np.loadtxt('paper_plots/pkin_wy20.txt')[2]

delta_pkin_y2_wy40 = np.loadtxt('paper_plots/pkin_wy40.txt')[1]
delta_pkin_z2_wy40 = np.loadtxt('paper_plots/pkin_wy40.txt')[2]

delta_pkin_y2_wy80 = np.loadtxt('paper_plots/pkin_wy80.txt')[1]
delta_pkin_z2_wy80 = np.loadtxt('paper_plots/pkin_wy80.txt')[2]

delta_pkin_y2_wy160 = np.loadtxt('paper_plots/pkin_wy160.txt')[1]
delta_pkin_z2_wy160 = np.loadtxt('paper_plots/pkin_wy160.txt')[2]
```

And plot

```python
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# ----------------------------
# Paper-style rcParams
# ----------------------------
mpl.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman", "Times", "STIXGeneral", "DejaVu Serif"],
    "mathtext.fontset": "stix",
    "font.size": 9,

    "axes.labelsize": 9,
    "axes.titlesize": 9,
    "axes.linewidth": 0.9,

    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.major.size": 5,
    "ytick.major.size": 5,
    "xtick.minor.size": 3,
    "ytick.minor.size": 3,
    "xtick.major.width": 0.9,
    "ytick.major.width": 0.9,
    "xtick.minor.width": 0.8,
    "ytick.minor.width": 0.8,
    
    "xtick.labelsize": 8,
    "ytick.labelsize": 8,
    "legend.fontsize": 8,

    "legend.frameon": False,
    "savefig.dpi": 300,
})

# ----------------------------
# Constants / helpers
# ----------------------------
shift = 50 / 1024 * 2
conv  = 2 / 0.197326

c_y, c_z = "tab:red", "tab:blue"

lw = 1.2
ms = 3.8
mew = 0.7

# Marker mapping for the simulation sets
markers = {10: "s", 20: "o", 40: "^", 80: "*", 160: "D"}
wvals = [0.3, 0.08, 0.02, 0.005, 0.001]
mseq  = ["s", "o", "^", "*", "D"]

# ----------------------------
# Figure
# ----------------------------
fig, ax = plt.subplots(figsize=(3.39, 2.6))   # nice single-column journal size

# y component
ax.plot(xp_direct + shift, delta_pkin_y2_wy10, linestyle="None", marker=markers[10],
        color=c_y, markersize=ms, markerfacecolor="none", markeredgewidth=mew)
ax.plot(xp_direct + shift, delta_pkin_y2_wy20, linestyle="None", marker=markers[20],
        color=c_y, markersize=ms, markerfacecolor="none", markeredgewidth=mew)
ax.plot(xp_direct + shift, delta_pkin_y2_wy40, linestyle="None", marker=markers[40],
        color=c_y, markersize=ms, markerfacecolor="none", markeredgewidth=mew)
ax.plot(xp_direct + shift, delta_pkin_y2_wy80, linestyle="None", marker=markers[80],
        color=c_y, markersize=ms, markerfacecolor="none", markeredgewidth=mew)
ax.plot(xp_direct + shift, delta_pkin_y2_wy160,  linestyle="None", marker=markers[160],
        color=c_y, markersize=ms, markerfacecolor="none", markeredgewidth=mew)

# z component
ax.plot(xp_direct + shift, delta_pkin_z2_wy10, linestyle="None", marker=markers[10],
        color=c_z, markersize=ms, markerfacecolor="none", markeredgewidth=mew)
ax.plot(xp_direct + shift, delta_pkin_z2_wy20, linestyle="None", marker=markers[20],
        color=c_z, markersize=ms, markerfacecolor="none", markeredgewidth=mew)
ax.plot(xp_direct + shift, delta_pkin_z2_wy40, linestyle="None", marker=markers[40],
        color=c_z, markersize=ms, markerfacecolor="none", markeredgewidth=mew)
ax.plot(xp_direct + shift, delta_pkin_z2_wy80, linestyle="None", marker=markers[80],
        color=c_z, markersize=ms, markerfacecolor="none", markeredgewidth=mew)
ax.plot(xp_direct + shift, delta_pkin_z2_wy160,  linestyle="None", marker=markers[160],
        color=c_z, markersize=ms, markerfacecolor="none", markeredgewidth=mew)

# classical lines
ax.plot(xp_class * conv, delta_pkin_y2_class / 3, linestyle="--", linewidth=lw, color=c_y)
ax.plot(xp_class * conv, delta_pkin_z2_class / 3, linestyle="--", linewidth=lw, color=c_z)

ax.set_xlabel(r"$x^+\ (\mathrm{GeV}^{-1})$")
ax.set_ylabel(r"$\langle (\delta p^{\mathrm{kin}}_{i})^2 \rangle \ (\mathrm{GeV}^2)$")
# ax.set_title(r"(a) dependence on $\langle (\delta r_y)^2 \rangle$")

ax.minorticks_on()
ax.grid(True, which="major", linewidth=0.5, alpha=0.25)
ax.grid(True, which="minor", linewidth=0.3, alpha=0.15)
ax.tick_params(top=True, right=True)
ax.set_xlim(left=0)
ax.margins(x=0.02, y=0.05)

# legends (width legend is 2-column)
legend_w_y = [
    Line2D([0],[0], marker=m, linestyle="None",
           markerfacecolor="none", markeredgewidth=mew,
           color="black", markersize=ms,
           label=fr"${w:.2g}$")
    for w, m in zip(wvals, mseq)
]
legend_w_y.append(Line2D([0],[0], linestyle="--", linewidth=lw, color="black", label="Classical"))

legend_comp = [
    Line2D([0],[0], color=c_y, lw=lw, label=r"$i=y$"),
    Line2D([0],[0], color=c_z, lw=lw, label=r"$i=z$"),
]

leg1 = ax.legend(handles=legend_w_y, loc="lower right",
                 title=r"$\sigma_y^2\ (\mathrm{GeV}^{-2})$",
                 ncol=2, columnspacing=1.2, handlelength=1.5,
                 handletextpad=0.6, labelspacing=0.35, borderpad=0.4)
ax.add_artist(leg1)
ax.legend(handles=legend_comp, loc="upper left", title="Component")

# ----------------------------
# Save
# ----------------------------
fig.tight_layout()
fig.savefig("paper_plots/y_width_dependence.pdf", bbox_inches="tight")
plt.show()
```

## 2. Jet quenching parameter

Load the results for $m_g=0$

```python
xplus_q15 = np.loadtxt('paper_plots/qhat_m=0_g2mu=1.33.txt')[0]
qhat_q15 = np.loadtxt('paper_plots/qhat_m=0_g2mu=1.33.txt')[1]

xplus_q2 = np.loadtxt('paper_plots/qhat_m=0_g2mu=2.txt')[0]
qhat_q2 = np.loadtxt('paper_plots/qhat_m=0_g2mu=2.txt')[1]

xplus_q25 = np.loadtxt('paper_plots/qhat_m=0_g2mu=2.5.txt')[0]
qhat_q25 = np.loadtxt('paper_plots/qhat_m=0_g2mu=2.5.txt')[1]
```

Plot the results in physical units

We plot the comparison between Coulomb and temporal gauges

```python
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# ----------------------------
# Paper-style rcParams
# ----------------------------
mpl.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman", "Times", "STIXGeneral", "DejaVu Serif"],
    "mathtext.fontset": "stix",
    "font.size": 9,

    "axes.labelsize": 9,
    "axes.titlesize": 9,
    "axes.linewidth": 0.9,

    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.major.size": 5,
    "ytick.major.size": 5,
    "xtick.minor.size": 3,
    "ytick.minor.size": 3,
    "xtick.major.width": 0.9,
    "ytick.major.width": 0.9,
    "xtick.minor.width": 0.8,
    "ytick.minor.width": 0.8,

    "xtick.labelsize": 8,
    "ytick.labelsize": 8,
    "legend.fontsize": 8,
    "legend.frameon": False,
    "savefig.dpi": 300,
})

# ----------------------------
# Constants
# ----------------------------
ms = 4.2
mew = 0.8
lw = 1.3

# Palette
colors = ["#D55E00", "#009E73", "#0072B2"]
markers = ["o", "s"]

# ----------------------------
# Figure
# ----------------------------
fig, ax = plt.subplots(figsize=(3.39, 2.6))

# ----------------------------
# Coulomb gauge results
# ----------------------------
ax.plot(xplus * 2.16, color_2_coulomb,
        linestyle="--", linewidth=lw,
        color=colors[0],
        # marker=markers[0], markersize=ms, markeredgewidth=mew,
        markerfacecolor="none")

ax.plot(xplus * 2.16, color_1_coulomb,
        linestyle="--", linewidth=lw,
        color=colors[1],
        # marker=markers[0], markersize=ms, markeredgewidth=mew,
        markerfacecolor="none")

ax.plot(xplus * 2.16, color_3_coulomb,
        linestyle="--", linewidth=lw,
        color=colors[2],
        # marker=markers[0], markersize=ms, markeredgewidth=mew,
        markerfacecolor="none")

# ----------------------------
# Temporal gauge results
# ----------------------------
ax.plot(xplus * 2.16, color_2_temp,
        linestyle="-", linewidth=lw,
        color=colors[0],
        # marker=markers[1], markersize=ms, markeredgewidth=mew,
        markerfacecolor="none")

ax.plot(xplus * 2.16, color_1_temp,
        linestyle="-", linewidth=lw,
        color=colors[1],
        # marker=markers[1], markersize=ms, markeredgewidth=mew,
        markerfacecolor="none")

ax.plot(xplus * 2.16, color_3_temp,
        linestyle="-", linewidth=lw,
        color=colors[2],
        # marker=markers[1], markersize=ms, markeredgewidth=mew,
        markerfacecolor="none")

# ----------------------------
# Reference line at 1/3
# ----------------------------
ax.axhline(1/3, color="0.3", linestyle=":", linewidth=1.6)
ax.text(0.12, 1/3 - 0.015,
        r"$1/3$",
        transform=ax.get_yaxis_transform(),
        va="top",
        ha="center",
        fontsize=8)

# ----------------------------
# Labels
# ----------------------------
ax.set_xlabel(r"$x^+ Q_s$")
ax.set_ylabel(r"$\langle P_c \rangle_{event}$")

# ----------------------------
# Ticks & grid
# ----------------------------
ax.minorticks_on()
ax.grid(True, which="major", linewidth=0.5, alpha=0.25)
ax.grid(True, which="minor", linewidth=0.3, alpha=0.15)
ax.tick_params(top=True, right=True)

ax.set_xlim(left=0, right=6)
ax.margins(x=0.02, y=0.05)

# ----------------------------
# Legends
# ----------------------------

# Style legend: dashed/open circle = Coulomb, solid/open square = Temporal
style_handles = [
    Line2D([0], [0],
           color="0.25", lw=lw,
           linestyle="--",
        #    marker=markers[0],
           markersize=ms,
           markerfacecolor="none",
           markeredgewidth=mew,
           label="Coulomb"),

    Line2D([0], [0],
           color="0.25", lw=lw,
           linestyle="-",
        #    marker=markers[1],
           markersize=ms,
           markerfacecolor="none",
           markeredgewidth=mew,
           label="Temporal"),
]

leg1 = ax.legend(handles=style_handles,
                 loc="center left",
                 bbox_to_anchor=(0.03, 0.52),
                 handlelength=2.0,
                 handletextpad=0.5,
                 borderpad=0.2,
                 labelspacing=0.35)

ax.add_artist(leg1)

# Color legend: dots only
color_handles = [
    Line2D([0], [0],
           color=colors[0],
        #    marker="o",
           linestyle="-",
           markersize=ms,
           markerfacecolor="none",
           markeredgewidth=mew,
           label=r"$c=1$"),

    Line2D([0], [0],
           color=colors[1],
        #    marker="o",
           linestyle="-",
           markersize=ms,
           markerfacecolor="none",
           markeredgewidth=mew,
           label=r"$c=2$"),

    Line2D([0], [0],
           color=colors[2],
        #    marker="o",
           linestyle="-",
           markersize=ms,
           markerfacecolor="none",
           markeredgewidth=mew,
           label=r"$c=3$"),
]

ax.legend(handles=color_handles,
          loc="upper right",
          bbox_to_anchor=(0.98, 0.8),
          handlelength=0.8,
          handletextpad=0.4,
          borderpad=0.2,
          labelspacing=0.35)

# ----------------------------
# Save
# ----------------------------
fig.tight_layout()
fig.savefig("paper_plots/color_rotation_gauge_comparison.pdf", bbox_inches="tight")
plt.show()
```
